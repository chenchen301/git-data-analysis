# FXI（凝血因子XI）数据分析
本 Notebook 使用整理后的教学数据，介绍 FXI 的结构、功能、缺乏症和研究方向。

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data.csv')
df
```

```python
print('各主题数量：')
print(df['topic'].value_counts())
```

```python
print('结构与功能信息：')
print(df[df['topic'].isin(['Structure', 'Function'])][['topic','indicator','value','unit']].to_string(index=False))
```

```python
summary = df.groupby('topic').size().sort_values(ascending=False)
summary.plot(kind='bar', title='FXI资料各主题数量')
plt.xlabel('主题')
plt.ylabel('条目数')
plt.tight_layout()
plt.show()
```

## 小结
FXI 是凝血系统中的重要蛋白。它可被 FXIIa 和凝血酶激活，活化后的 FXIa 能激活 FIX。FXI 缺乏可导致血友病C，并常表现为程度不一的出血倾向。需要注意：本项目中的表格是用于编程练习的教学整理数据，不是患者临床数据。
