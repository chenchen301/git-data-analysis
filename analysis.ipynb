# 学生成绩数据分析

本 Notebook 用于展示一个可复现的 Git + Python 数据分析项目。

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

DATA_FILE = Path('data.csv')
df = pd.read_csv(DATA_FILE)
df
```

```python
subjects = ['数学', '英语', 'Python']
df['总分'] = df[subjects].sum(axis=1)
df['平均分'] = df[subjects].mean(axis=1).round(2)
df[['姓名', *subjects, '总分', '平均分']]
```

```python
summary = df[subjects].agg(['mean', 'max', 'min']).T.round(2)
summary.columns = ['平均分', '最高分', '最低分']
summary
```

```python
print('班级平均分：', round(df['平均分'].mean(), 2))
top = df.loc[df['平均分'].idxmax()]
print('平均分最高：', top['姓名'], top['平均分'])
print('\n成绩等级人数：')
print(pd.cut(df['平均分'], bins=[0,70,80,90,100], labels=['D','C','B','A'], right=False).value_counts().sort_index())
```

```python
df[subjects].mean().plot(kind='bar', figsize=(7,4), title='各科平均成绩')
plt.xlabel('科目')
plt.ylabel('平均分')
plt.tight_layout()
plt.show()
```

```python
df['平均分'].plot(kind='hist', bins=5, figsize=(7,4), title='学生平均分分布')
plt.xlabel('平均分')
plt.ylabel('人数')
plt.tight_layout()
plt.show()
```
